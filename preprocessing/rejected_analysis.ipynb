```python
import pandas as pd

def generate_sanity_check(rejected_path, accepted_path, n_samples=50):
    df_rejected = pd.read_csv(rejected_path)
    df_accepted = pd.read_csv(accepted_path, sep='\t')

    sample_rejected = df_rejected.sample(n=min(len(df_rejected), n_samples), random_state=42)
    sample_accepted = df_accepted.sample(n=min(len(df_accepted), n_samples), random_state=42)

    sample_rejected['manual_validation'] = ""
    sample_accepted['manual_validation'] = ""
    
    sample_rejected['stage3_decision'] = 'REJECTED'
    sample_accepted['stage3_decision'] = 'ACCEPTED'

    audit_df = pd.concat([sample_rejected, sample_accepted], ignore_index=True)
    
    output_path = "audit_stage3_bart.csv"
    audit_df.to_csv(output_path, index=False)
    print(f"Arquivo de auditoria gerado: {output_path}")
    return audit_df

generate_sanity_check('rejecteds/03_bart_rejected_samples.csv', 'processed/usnid/train.tsv')
```
